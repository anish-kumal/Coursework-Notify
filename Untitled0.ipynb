```python
import folium
import geopandas as gpd
from shapely.geometry import Point, Polygon
from shapely.ops import transform
import pyproj

# KATHMANDU BOUNDING BOX
min_lat, max_lat = 27.60, 27.80
min_lon, max_lon = 85.20, 85.45

ktm_polygon = Polygon([
    (min_lon, min_lat),
    (max_lon, min_lat),
    (max_lon, max_lat),
    (min_lon, max_lat),
    (min_lon, min_lat),
]) #polygon has 4 points but during processing we take the start and end point


def inside_box(coords):
    lat, lon = coords
    return ktm_polygon.contains(Point(lon, lat))

# TEMPLES DATA
temples = [
    {"name": "Pashupatinath Temple",         "coords": [27.7105, 85.3487]},
    {"name": "Swayambhunath Stupa","coords": [27.7149, 85.2904]},
    {"name": "Changu Narayan Temple",        "coords": [27.7308, 85.4190]},
    {"name": "Dakshinkali Temple",           "coords": [27.5963, 85.2018]},  # outside box or may be outlier
]



#  FUNCTION: creating 500m buffer in meters
def make_buffer(lat, lon, radius_m=500):

    wgs84 = pyproj.CRS("EPSG:4326")       # lat/lon coordinate system
    utm   = pyproj.CRS("EPSG:32645")      # UTM Zone 45N (covering Nepal)

    project_to_utm   = pyproj.Transformer.from_crs(wgs84, utm,   always_xy=True).transform
    project_to_wgs84 = pyproj.Transformer.from_crs(utm,   wgs84, always_xy=True).transform

    point_wgs84 = Point(lon, lat)
    point_utm   = transform(project_to_utm,   point_wgs84)   # now in meters
    buffer_utm  = point_utm.buffer(radius_m)                 # 500m circle in meters
    buffer_wgs84 = transform(project_to_wgs84, buffer_utm)   # back to lat/lon

    return buffer_wgs84

# CREATE MAP
ktm_map = folium.Map(location=[27.7172, 85.3240], zoom_start=12)

# Draw bounding rectangle
folium.Rectangle(
    bounds=[[min_lat, min_lon], [max_lat, max_lon]],
    color='red', weight=2,
    fill=True, fill_color='lightyellow', fill_opacity=0.05,
    tooltip="Kathmandu Bounding Box"
).add_to(ktm_map)

ktm_map.fit_bounds([[min_lat, min_lon], [max_lat, max_lon]])

# LAYERS
marker_layer = folium.FeatureGroup(name="Temples")
buffer_layer = folium.FeatureGroup(name="500m Buffer Zones")

# ADDING MARKERS + BUFFERS (only inside bounding box)
for t in temples:
    if not inside_box(t["coords"]):
        print(f"Skipped (outside box): {t['name']}")
        continue

    lat, lon = t["coords"]

    # Temple marker
    folium.Marker(
        location=[lat, lon],
        popup=folium.Popup(f"<b>{t['name']}</b>", max_width=200),
        tooltip=t["name"],
        icon=folium.Icon(color='red', icon='star', prefix='fa')
    ).add_to(marker_layer)

    # 500m buffer circle (true meters, not degrees)
    buffer_geom = make_buffer(lat, lon, radius_m=500)

    folium.GeoJson(
        buffer_geom.__geo_interface__,
        style_function=lambda x: {
            "fillColor": "#ff4444 ",
            "color":     "#00008B",
            "weight":    2,
            "fillOpacity": 0.15,
            "dashArray": "5, 5"
        },
        tooltip=f"{t['name']} — 500m buffer"
    ).add_to(buffer_layer)

marker_layer.add_to(ktm_map)
buffer_layer.add_to(ktm_map)
folium.LayerControl().add_to(ktm_map)

ktm_map
```
